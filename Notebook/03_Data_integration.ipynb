```python
import pandas as pd
```

```python
from pathlib import Path

BASE = Path("../Data/Processed/wellness")

fatigue = pd.read_csv(BASE / "fatigue_long.csv")
mood = pd.read_csv(BASE / "mood_long.csv")
readiness = pd.read_csv(BASE / "readiness_long.csv")
sleep_duration = pd.read_csv(BASE / "sleep_duration_long.csv")
sleep_quality = pd.read_csv(BASE / "sleep_quality_long.csv")
soreness = pd.read_csv(BASE / "soreness_long.csv")
stress = pd.read_csv(BASE / "stress_long.csv")
```

```python
wellness_master = fatigue
```

```python
wellness_master = wellness_master.merge(
    mood,
    on=["date", "Player Name"],
    how="outer"
)
```

```python
wellness_master = wellness_master.merge(
    readiness,
    on=["date", "Player Name"],
    how="outer"
)
```

```python
wellness_master = wellness_master.merge(
    sleep_duration,
    on=["date", "Player Name"],
    how="outer"
)
```

```python
wellness_master = wellness_master.merge(
    sleep_quality,
    on=["date", "Player Name"],
    how="outer"
)
```

```python
wellness_master = wellness_master.merge(
    soreness,
    on=["date", "Player Name"],
    how="outer"
)
```

```python
wellness_master = wellness_master.merge(
    stress,
    on=["date", "Player Name"],
    how="outer"
)
```
