```python
logger.info("Importing libraries Like Pandas and StandardScaler")
import pandas as pd
from pathlib import Path
from sklearn.preprocessing import StandardScaler

logger.info("Importing libraries Like Pandas and StandardScaler Completed Successfully")

# Resolve dataset path from notebook environment
root = Path.cwd()
search_paths = [
    root / "Data" / "processed" / "master_dataset.csv",
    root.parent / "Data" / "processed" / "master_dataset.csv",
]
dataset_path = None
for path in search_paths:
    if path.exists():
        dataset_path = path
        break

if dataset_path is None:
    candidates = list(root.rglob("master_dataset.csv"))
    if candidates:
        dataset_path = candidates[0]
    else:
        raise FileNotFoundError(
            f"Could not find master_dataset.csv in {root} or its parents."
        )

# Load merged dataset
df = pd.read_csv(dataset_path)

# Continuous variables
continuous_cols = [
    "distance",
    "duration",
    "high_speed_distance",
    "low_speed_distance",
    "avg_speed",
    "max_speed",
    "accelerations",
    "decelerations",
    "player_load"
]

# Create scaler
scaler = StandardScaler()

# Normalize columns
logger.info("Normalizing continuous columns")
df[continuous_cols] = scaler.fit_transform(df[continuous_cols])

# Save processed data
logger.info("Saving processed data")
output_path = dataset_path.parent / "master_dataset_normalized.csv"
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)

print("Normalization completed.")
```
