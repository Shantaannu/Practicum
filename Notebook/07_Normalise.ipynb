```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load merged dataset
df = pd.read_csv("Data/processed/master_dataset.csv")

# Continuous variables
continuous_cols = [
    "distance",
    "duration",
    "high_speed_distance",
    "low_speed_distance",
    "avg_speed",
    "max_speed",
    "accelerations",
    "decelerations",
    "player_load"
]

# Create scaler
scaler = StandardScaler()

# Normalize columns
df[continuous_cols] = scaler.fit_transform(df[continuous_cols])

# Save processed data
df.to_csv(
    "Data/processed/master_dataset_normalized.csv",
    index=False
)

print("Normalization completed.")
```
