```python
import pandas as pd
```

```python
import os
print(os.getcwd())
```

```python

fatigue_df = pd.read_csv("../Data/subjective/wellness/fatigue.csv")
fatigue_df.head()
```

```python
print(fatigue_df.columns[:10])
```

```python
def reshape_wide_to_long(df, value_name):
    id_col = df.columns[0]
    return (
        df.melt(id_vars=[id_col], var_name="Player Name", value_name=value_name)
          .rename(columns={id_col: "date"})
    )

fatigue_long = reshape_wide_to_long(fatigue_df, "fatigue")
```

```python
from pathlib import Path

out_path = Path("../Data/Processed/wellness/fatigue_long.csv")

out_path.parent.mkdir(parents=True, exist_ok=True)
fatigue_long.to_csv(out_path, index=False)

print(out_path.resolve())
```
